# 3d visualization of the examples using Plotly

This notebook uses the Plotly library to simulate 3D models of the test models provided in the Input_files/ folder, and interact with them. Each element of the legend can be hidden/shown. Outputs are also needed so they must have been created before using this notebook. To generate them, follow the README.md. The outputs will be stocked in the outputs/ folder.

August. 2026

```python
import plotly.graph_objects as go
import pandas as pd
import notebook_functions as nf
```

## 1) Base simulation
A single-phase run conducted on a synthetic dataset, illustrating how simulations are directed by :
- vadose/phreatic partitioning,
- fracture guidance (two families),
- intrinsic karstification potential,
- inception surfaces.

The 3D viewer shows several elements such as inception surfaces, water table surfaces, the topographic surface, inlets, outlets and the base simulation.

```python
## Import data
# Inputs : they are imported and displayed directly using the add_inputs function of notebook_functions.py
# add_inputs contains the inlets, outlets, waypoints, box limits, karst-free spheres, water table surfaces and inception surfaces, and add them to the figure

# Outputs
sim = pd.read_csv('outputs/base_0_karst.txt', delimiter = '\t')  # Simulation points of the base simulation


## Data extraction
# Simulation data separated into several segments created in the simulation so that everything can be shown properly with "lines" and put in a single legend
x_base, y_base, z_base = nf.separate_segment(sim)  # Base simulation


## Add the elements to the figure
# Figure creation
fig = go.Figure()

# Outputs
fig.add_trace(go.Scatter3d(x=x_base, y=y_base, z=z_base, mode='lines+markers', line=dict(color='green', width=15), marker=dict(color='green', size=3), name="Base simulation"))  # Base simulation

# Inputs
nf.add_inputs(fig)  # Shows all the inputs

# Layout + show
fig.update_layout(title = '3D visualization of the base simulation', height = 850, width = 1100, legend=dict(x=0.8, y=0.95, bgcolor='rgba(255,255,255,0.5)'), scene=dict(xaxis_title="X", yaxis_title="Y", zaxis_title="Z"))  # Legend
fig.show()
```

## 2) Polygenic karst (multi-phase reuse)
Reuses a previously generated network by down-weighting already traversed edges, mimicking multi-phase karstification (here, base-level drop with flow redirected to another outlet). The example disables the fracture term to emulate a more homogeneous medium in the bottom of the aquifer, and showcases waypoints and karst-free points close to the aquifer substratum.

The 3D viewer shows elements that are the same as in the previous simulation, with the addition of the polyphasic simulation.

```python
## Import data
# Outputs
sim_pol = pd.read_csv('outputs/polyphasic_0_karst.txt', delimiter = '\t')  # Simulation points of the polyphasic simulation


## Data extraction
# Simulation data separated into several segments created in the simulation so that everything can be shown properly with "lines" and put in a single legend
x_pol, y_pol, z_pol = nf.separate_segment(sim_pol)  # Polyphasic simulation


## Add the elements to the figure
# Figure creation
fig = go.Figure()

# Outputs
fig.add_trace(go.Scatter3d(x = x_base, y = y_base, z = z_base, mode='lines+markers', line=dict(color='green', width=15), marker=dict(color='green', size=3), name="Base simulation"))  # Base simulation
fig.add_trace(go.Scatter3d(x = x_pol, y = y_pol, z = z_pol, mode='lines+markers', line=dict(color='magenta', width=15), marker=dict(color='magenta', size=3), name="Polyphasic simulation"))  # Polyphasic simulation

# Inputs
nf.add_inputs(fig)

# Layout + show
fig.update_layout(title = '3D visualization of the polyphasic simulation', height = 850, width = 1100, legend=dict(x=0.83, y=0.95, bgcolor='rgba(255,255,255,0.5)'), scene=dict(xaxis_title="X", yaxis_title="Y", zaxis_title="Z"))  # Legend
fig.show()
```

## 3) Amplification
Starting from the base network, this example runs the amplification step to control density and topology by adding :
- small-scale dead-end branches,
- a limited number of cycles/loops.
  
The 3D viewer shows elements that are the same as in the first simulation, but with the polyphasic and amplificated simulation instead of the base simulation.

```python
## Import data
# Outputs
sim_pol_amp = pd.read_csv('outputs/amplification_0_karst.txt', delimiter = '\t')  # Simulation points of the polyphasic and amplificated simulation 


## Data extraction
# Simulation data separated into several segments created in the simulation so that everything can be shown properly with "lines" and put in a single legend
x_pol_amp, y_pol_amp, z_pol_amp = nf.separate_segment(sim_pol_amp)  # Polyphasic and amplificated simulation 


## Add the elements to the figure
# Figure creation
fig = go.Figure()

# Outputs
fig.add_trace(go.Scatter3d(x = x_pol_amp, y = y_pol_amp, z = z_pol_amp, mode='lines+markers', line=dict(color='black', width=15), marker=dict(color="black", size=3), name="Polyphasic and amplificated simulation"))

# Inputs
nf.add_inputs(fig)  

## Layout + show
fig.update_layout(title = '3D visualization of the polyphasic and amplificated simulation', height = 850, width = 1100, legend=dict(x=0.70, y=0.95, bgcolor='rgba(255,255,255,0.5)'), scene=dict(xaxis_title="X", yaxis_title="Y", zaxis_title="Z"))  # Legend
fig.show()
```

## 4) Karst section generation (curvilinear SGS)
Given the amplified network, this example simulates equivalent radii on nodes using a curvilinear SGS algorithm (Frantz et al., 2021). Variogram parameters mirror those in the paper.

The 3D viewer shows elements that are the same as in the latest simulation, but with the generated karst section instead of the polyphasic and amplificated simulation.

```python
## Import data
# Outputs
gen_sec = pd.read_csv('outputs/section_gen_0_karst.txt', delimiter = '\t')    # Simulation points of the generated section using the polyphasic and amplificated simulation


## Add the elements to the figure
# Figure creation
fig = go.Figure()

# Outputs
fig.add_trace(go.Scatter3d(x = gen_sec.X, y = gen_sec.Y, z = gen_sec.Z, mode='markers', marker=dict(color="grey", size=abs(gen_sec.equivalent_radius)*10), name="Generated karst section"))

# Inputs
nf.add_inputs(fig)

## Layout + show
fig.update_layout(title = '3D visualization of the generated karst section', height = 850, width = 1100, legend=dict(x=0.82, y=0.95, bgcolor='rgba(255,255,255,0.5)'), scene=dict(xaxis_title="X", yaxis_title="Y", zaxis_title="Z"))  # Legend
fig.show()
```
